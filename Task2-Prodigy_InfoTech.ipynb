```python
import pandas as pd

url = "https://raw.githubusercontent.com/Prodigy-InfoTech/data-science-datasets/main/Task%202/train.csv"
df = pd.read_csv(url)

df.head()
```

```python
#Step 3: Understand the Data
df.info()
df.head()
```

```python
df.describe(include='all')
```

```python
print("Shape:", df.shape)
print("Columns:", df.columns.tolist())
```

```python
#Step 4: Data Cleaning
```

```python
df.isnull().sum()
```

```python
df['Age'] = df['Age'].fillna(df['Age'].median())
df['Embarked'] = df['Embarked'].fillna(df['Embarked'].mode()[0])

```

```python
df.drop_duplicates(inplace=True)
```

```python
df.dtypes
```

```python
df['Survived'] = df['Survived'].astype('category')
df['Pclass'] = df['Pclass'].astype('category')
```

```python
#Step 5: Exploratory Data Analysis (EDA)
```

```python
#1. Univariate Analysis
df['Age'].hist(bins=30)
plt.title('Age Distribution')
plt.show()
```

```python
sns.countplot(x='Survived', data=df)
plt.title('Survival Count')
plt.show()
```

```python
#2. Bivariate Analysis
sns.countplot(x='Sex', hue='Survived', data=df)
plt.title('Survival by Gender')
plt.show()
```

```python
sns.countplot(x='Pclass', hue='Survived', data=df)
plt.title('Survival by Passenger Class')
plt.show()
```

```python
sns.boxplot(x='Survived', y='Age', data=df)
plt.title('Age Distribution by Survival')
plt.show()
```

```python
#3. Multivariate Analysis
sns.pairplot(df, hue='Survived', vars=['Age', 'Fare'])
plt.show()
```

```python
corr = df.corr(numeric_only=True)
sns.heatmap(corr, annot=True, cmap='coolwarm')
plt.title('Correlation Heatmap')
plt.show()
```

```python
#step 6: Identify Patterns and Trends
Examples of insights we might find:
Age: Younger passengers had slightly higher survival.
Fare: Higher fare often correlates with higher survival.
```
